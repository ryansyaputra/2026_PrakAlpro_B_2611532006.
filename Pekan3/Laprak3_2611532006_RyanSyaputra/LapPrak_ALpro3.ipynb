## 2.1 Dasar Teori


Python merupakan bahasa pemrograman tingkat tinggi (high-level language) yang bersifat interpreted, berorientasi objek, serta mengusung mekanisme dynamic typing. Bahasa yang dikembangkan oleh Guido van Rossum ini populer karena memiliki struktur sintaks yang sederhana, intuitif, dan mudah dibaca, sehingga ideal untuk mempelajari dasar pemrograman hingga membangun sistem kompleks seperti kecerdasan buatan (artificial intelligence). Dalam struktur program Python, variabel berperan sebagai wadah penyimpanan nilai sementara di memori agar data dapat diakses dan diolah kembali saat program dijalankan. Pembuatan variabel dilakukan secara langsung melalui penugasan nilai (assignment) menggunakan operator = tanpa deklarasi tipe data eksplisit, dengan aturan penulisan yang bersifat case-sensitive, tidak diawali angka, dan menghindari penggunaan keyword bawaan Python.

Setiap nilai yang disimpan di dalam variabel diwakili oleh tipe data yang menentukan jenis operasi komputasi yang dapat dilakukan. Beberapa tipe data dasar pada Python meliputi integer (int) untuk bilangan bulat, float (float) untuk bilangan desimal, string (str) untuk teks atau karakter yang diapit tanda petik, serta boolean (bool) untuk nilai logika True atau False. Melalui fitur dynamic typing, Python secara otomatis mengenali tipe data variabel berdasarkan nilai yang dimasukkan. Namun, dalam situasi tertentu diperlukan proses type casting atau konversi tipe data manual—seperti penggunaan fungsi int(), float(), atau str()—untuk memastikan ketersesuaian tipe data dan mencegah terjadinya kesalahan (type error) saat eksekusi program.

### 2.2.2 Program "Numerik_2611532006"


```python
# Buat file dengan nama Numerik_NIM.py
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
#create a variable with integer value.
a_2006 = 100
print("The type of variable having value", a_2006, "is", type(a_2006))

#create a variable with float value
b_2006 = 10.2345
print("The type of variable having value", b_2006, "is", type(b_2006))

#create a variable with complex value
c_2006 = 100+3j
print("The type of variable having value", c_2006, "is", type(c_2006))
```

Complex (complex): Baris ke-12 mendefinisikan variabel c_2006 dengan nilai 100+3j, yang merupakan bilangan kompleks yang terdiri dari bagian riil (100) dan bagian imajiner (3j).

### 2.2.3 Program "Konstanta_2611532006"


```python
# Buat file dengan nama Konstanta_NIM.py
# program ini menggunakan konstanta untuk menghitung luas lingkaran
# Nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI : Final = 3.14
print("pi: %f" % (PI))
jari_2006 = float(input('Masukkan nilai jari-jari:'))
luas_2006 = PI * jari_2006 * jari_2006
print("Luas lingkaran dengan jari-jari %.2f adalah %2f" % (jari_2006, luas_2006))
```

**Penjelasan :**
1. Modul Final diimpor dari library typing untuk menegaskan bahwa variabel PI bersifat konstanta (nilainya bernilai tetap 3.14 dan tidak boleh diubah).

2. Fungsi input() digunakan untuk menerima masukan nilai jari-jari dari pengguna. Nilai tersebut kemudian diubah (type casting) menjadi tipe data pecahan menggunakan fungsi float() dan disimpan ke dalam variabel jari_2006.

3. Baris ke-9 melakukan perhitungan luas lingkaran menggunakan rumus $\text{PI} \times r^2$ yang disimpan pada variabel luas_2006.Fungsi print() memanfaatkan string formatting (operator %) untuk menampilkan nilai keluaran, seperti %.2f yang membatasi tampilan desimal hingga dua angka di belakang koma.

### 2.2.5 Program "Boolean_2611532006"


```python
#Buat file dengan nama Boolean_NIM.py
#Nama variabel ditambah 4 digit nim terakhir contoh: milai_1234
#Deklarasi variabel dengan tipe data boolean
is_lulus_2006 = True
is_cumlaude_2006 = True

#Menggunakan Boolean
nilai_2006 = 85
batas_lulus_2006 = 75

#menentukan nilai boolean dari kondisi
status_kelulusan_2006 = nilai_2006 >= batas_lulus_2006 #hasilnya akan true

print("=== Check kelulusan ===")
print("Nilai:",nilai_2006 )
print("Apakah lulus?:", status_kelulusan_2006)
if is_lulus_2006 and is_cumlaude_2006:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

2. Baris 12 melakukan perbandingan menggunakan operator relasional >= (greater than or equal to). Karena nilai 85 lebih besar dari 75, ekspresi nilai_2006 >= batas_lulus_2006 menghasilkan nilai True yang kemudian disimpan di dalam variabel status_kelulusan_2006.

3. Baris 14–16 menampilkan header teks, nilai mahasiswa, serta hasil evaluasi boolean (True) ke terminal layar.
