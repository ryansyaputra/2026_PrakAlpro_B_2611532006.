## 2.1 Dasar Teori


1. Conditional pada Python

Conditional atau percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat keputusan berdasarkan kondisi tertentu. Dengan conditional, program dapat menentukan perintah yang akan dijalankan sesuai dengan kondisi atau nilai yang diberikan.

Dalam bahasa pemrograman Python, percabangan dapat dilakukan menggunakan beberapa struktur, yaitu if, if-else, if-elif-else, multi if, dan match-case.

2. If

if merupakan struktur percabangan yang digunakan untuk memeriksa suatu kondisi. Perintah di dalam if akan dijalankan apabila kondisi yang diperiksa bernilai benar atau True. Jika kondisi bernilai salah atau False, maka perintah tersebut tidak akan dijalankan.

3. Multi If

Multi if merupakan penggunaan beberapa pernyataan if secara terpisah dalam satu program. Setiap kondisi pada masing-masing if diperiksa secara independen. Oleh karena itu, lebih dari satu kondisi dapat terpenuhi dan dijalankan dalam satu proses.

4. If-Else

if-else merupakan struktur percabangan yang digunakan ketika terdapat dua kemungkinan berdasarkan suatu kondisi. Jika kondisi pada if terpenuhi, maka perintah pada bagian if dijalankan. Sebaliknya, jika kondisi tidak terpenuhi, maka perintah pada bagian else akan dijalankan.

5. If-Elif-Else

if-elif-else digunakan ketika terdapat beberapa kemungkinan kondisi yang harus diperiksa. if digunakan untuk memeriksa kondisi pertama, sedangkan elif digunakan untuk memeriksa kondisi lainnya apabila kondisi sebelumnya tidak terpenuhi. Bagian else digunakan apabila seluruh kondisi sebelumnya tidak terpenuhi.

Pemeriksaan kondisi dilakukan secara berurutan. Setelah ditemukan kondisi yang bernilai benar, perintah pada kondisi tersebut dijalankan dan kondisi berikutnya tidak diperiksa.

6. Match-Case

match-case merupakan struktur percabangan pada Python yang digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan atau pola tertentu. Struktur ini dapat digunakan ketika sebuah variabel memiliki beberapa kemungkinan nilai yang masing-masing membutuhkan tindakan berbeda.

Pada match-case, setiap pilihan dituliskan menggunakan case. Python akan mencocokkan nilai yang diperiksa dengan setiap case. Jika tidak terdapat pilihan yang sesuai, case _ dapat digunakan sebagai kondisi default.

## 2.2 Kode Program

### 2.2.1 Program "if1_2611532006"


```python
# Buat file dengan nama if1_2611532006.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2006
# Program ini menggunakan fungsi input()

ipk_2006 = float(input("Input IPK Anda: "))

if ipk_2006 > 2.75:
    print("Anda lulus sangat memuaskan dengan IPK" + str(ipk_2006))

print("Program Selesai") 
```

**Penjelasan :**
1. Deklarasi variabel dan input.
Program menggunakan fungsi input() untuk meminta pengguna memasukkan nilai IPK. Nilai tersebut kemudian dikonversi menjadi tipe data float karena IPK dapat memiliki nilai berupa bilangan desimal dan disimpan ke dalam variabel ipk_2006.
2. Percabangan if.
Program memeriksa apakah nilai ipk_2006 lebih besar dari 2,75. Jika kondisi tersebut terpenuhi, program akan menampilkan pesan bahwa pengguna lulus sangat memuaskan beserta nilai IPK yang dimasukkan.
3. Konversi tipe data.
Nilai IPK yang awalnya berupa tipe data float diubah menjadi string menggunakan str() agar dapat digabungkan dengan teks pada fungsi print().
4. Menampilkan akhir program.
Setelah proses percabangan selesai, program menampilkan tulisan "Program Selesai". Perintah ini tetap dijalankan meskipun kondisi if tidak terpenuhi karena berada di luar blok if.


### 2.2.2 Program "if2_2611532006"


```python
# Buat file dengan nama if2_2611532006.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2006
# Program ini menggunakan fungsi input()

ipk_2006 = float(input("Input IPK Anda: "))

if ipk_2006 > 2.75:
    print("Anda lulus sangat memuaskan dengan IPK" + str(ipk_2006))
else:
    print("Anda Tidak Lulus") 
print("Program Selesai")
```

**Penjelasan :**
1. `ipk_2006 = float(input("Input IPK Anda: "))` digunakan untuk meminta pengguna memasukkan nilai IPK. Nilai input diubah menjadi tipe data `float` agar dapat digunakan sebagai bilangan desimal.

2. `if ipk_2006 > 2.75:` digunakan untuk memeriksa apakah nilai IPK lebih besar dari 2,75. Jika kondisi terpenuhi, program akan menjalankan perintah di dalam `if`.

3. `print("Anda lulus sangat memuaskan dengan IPK" + str(ipk_2006))` digunakan untuk menampilkan pesan bahwa pengguna lulus sangat memuaskan beserta nilai IPK. Fungsi `str()` digunakan untuk mengubah nilai IPK menjadi teks agar dapat digabungkan dengan kalimat.

4. `else:` digunakan ketika kondisi pada `if` tidak terpenuhi.

5. `print("Anda Tidak Lulus")` digunakan untuk menampilkan pesan "Anda Tidak Lulus" apabila nilai IPK tidak lebih besar dari 2,75.

6. `print("Program Selesai")` digunakan untuk menampilkan pesan bahwa program telah selesai dijalankan. Perintah ini berada di luar blok `if-else`, sehingga tetap dijalankan pada kedua kondisi.


### 2.2.3 Program "if_elif_else1_2611532006"


```python
# Buat file dengan nama if_elif_else1_2611532006.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2006
# Program ini menggunakan fungsi input()

umur_2006 = int(input("Input umur anda: "))
sim_2006 = input("Apakah Anda Sudah Punya Sim C: ")[0]

if umur_2006 >= 17 and sim_2006 == 'y':
    print("Anda Sudah dewasa dan boleh bawa motor")
elif umur_2006 >= 17 and sim_2006 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh bawa motor")
elif umur_2006 < 17 and sim_2006 == 'y':
    print("Anda Belum Cukup Umur punya SIM")
else:
    print("Anda Belum Cukup Umur dan tidak boleh bawa motor")

print("Program Selesai")
```

**Penjelasan :**
1. `umur_2006 = int(input("Input umur anda: "))` digunakan untuk meminta pengguna memasukkan umur. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat.

2. `sim_2006 = input("Apakah Anda Sudah Punya Sim C: ")[0]` digunakan untuk menerima input mengenai kepemilikan SIM C. `[0]` digunakan untuk mengambil karakter pertama dari input.

3. `if umur_2006 >= 17 and sim_2006 == 'y':` digunakan untuk memeriksa apakah pengguna sudah berusia minimal 17 tahun dan memiliki SIM C. Operator `and` digunakan jika kedua kondisi harus terpenuhi.

4. `elif` digunakan untuk memeriksa kondisi lain apabila kondisi sebelumnya tidak terpenuhi. Pada program ini terdapat dua kondisi `elif`, yaitu kondisi pengguna sudah berusia minimal 17 tahun tetapi tidak memiliki SIM C, serta kondisi pengguna belum berusia 17 tahun tetapi memiliki SIM C.Ketika kode di run, `elif` yang pertama true, maka `elif` yang setelah nya tidak akan dicek kebenaran nya. 

5. `else:` digunakan apabila seluruh kondisi pada `if` dan `elif` tidak terpenuhi.

6. `print()` digunakan untuk menampilkan hasil sesuai dengan kondisi yang terpenuhi. Program memiliki pesan yang berbeda untuk setiap kondisi.

7. `print("Program Selesai")` digunakan untuk menampilkan pesan bahwa program telah selesai dijalankan.


### 2.2.4 Program "match_case1_2611532006"


```python
# Buat file dengan nama match_case1_2611532006.py
# Buat program untuk match case
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_2006
# Program ini menggunakan fungsi input()
# program konversi angka menjadi nama bulan

bulan_2006 = int(input("Masukkan angka bulan (1 - 12): "))

match bulan_2006:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. `bulan_2006 = int(input("Masukkan angka bulan (1 - 12): "))` digunakan untuk meminta pengguna memasukkan angka bulan. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat.

2. `match bulan_2006:` digunakan untuk mencocokkan nilai yang terdapat pada variabel `bulan_2006` dengan pilihan yang tersedia pada setiap `case`.

3. `case 1` sampai `case 12` digunakan untuk mencocokkan angka bulan dengan nama bulan yang sesuai, mulai dari Januari hingga Desember.

4. `case _:` digunakan sebagai kondisi default apabila nilai yang dimasukkan tidak sesuai dengan `case` yang tersedia.

5. `print()` digunakan untuk menampilkan nama bulan sesuai dengan angka yang dimasukkan atau menampilkan pesan "Angka tidak valid" jika input berada di luar angka 1 sampai 12.


### 2.2.5 Program "multi_if1_2611532006"

### 2.2.6 Program "multi_if2_2611532006"


```python
# Buat file dengan nama multi_if2_2611532006.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_2006
# Program ini menggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2006 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2006 = input("Apakah Anda member? (y/t): ").strip().lower()
is_member = input_member_2006 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2006 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_valid_2006 = input_promo_2006 in ["y", "ya"]

total_diskon_persen_2006 = 0

# Multi-IF terpisah: Setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2006 >= 1000000:
    total_diskon_persen_2006 += 10 # Diskon belanja besar

if is_member:
    total_diskon_persen_2006 += 5 # Diskon member

if kode_promo_valid_2006:
    total_diskon_persen_2006 += 15 # Diskon voucher

# Menghitung nomimal diskon dan total bayar
nominal_diskon_2006 = total_belanja_2006 * (total_diskon_persen_2006 / 100)
total_bayar_2006 = total_belanja_2006 - nominal_diskon_2006

#Output hasil
print("\n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_2006}% (Rp {nominal_diskon_2006:,.0f})")
print(f"Total Bayar  : Rp {total_bayar_2006:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_2006}%")
# Putput total diskon yang anda dapatkan: 30% jika belanja > 1juta, member, dan kode promo valid
```

1. `total_belanja_2006 = float(input(...))` digunakan untuk menerima input total belanja dari pengguna. Fungsi `float()` digunakan agar nilai dapat berupa bilangan desimal.

2. `input_member_2006 = input(...).strip().lower()` digunakan untuk menerima status member. `strip()` menghapus spasi di awal dan akhir input, sedangkan `lower()` mengubah huruf menjadi huruf kecil.

3. `is_member = input_member_2006 in ["y", "ya"]` digunakan untuk memeriksa apakah input pengguna berupa `y` atau `ya`. Hasil pemeriksaan berupa `True` atau `False`.

4. `kode_promo_valid_2006` digunakan dengan cara yang sama untuk menentukan apakah kode promo yang dimasukkan valid.

5. `total_diskon_persen_2006 = 0` digunakan untuk memberikan nilai awal total diskon sebesar 0 persen sebelum kondisi diperiksa.

6. Program menggunakan beberapa `if` yang berdiri sendiri sehingga setiap kondisi diperiksa tanpa memedulikan hasil if lainnya. Diskon dapat ditambahkan apabila pengguna memenuhi beberapa syarat sekaligus.

7. Jika total belanja minimal Rp1.000.000, maka diskon ditambah 10%.

8. Jika pengguna merupakan member, maka diskon ditambah 5%.

9. Jika kode promo valid, maka diskon ditambah 15%.

10. `nominal_diskon_2006` digunakan untuk menghitung jumlah uang yang diperoleh dari total persentase diskon.

11. `total_bayar_2006` digunakan untuk menghitung jumlah yang harus dibayar setelah dikurangi nominal diskon.

12. `print()` digunakan untuk menampilkan rincian pembayaran, yaitu total diskon dalam persen, nominal diskon dalam rupiah, dan total pembayaran.

13. Jika ketiga kondisi terpenuhi, total diskon akan terakumulasi menjadi 30%, karena masing-masing diskon 10%, 5%, dan 15% dijumlahkan.

1. Conditional (percabangan) memungkinkan program mengambil keputusan dan menjalankan perintah tertentu sesuai kondisi yang bernilai True atau False, sehingga alur program tidak selalu berjalan lurus dari atas ke bawah.

2. Pernyataan if menjalankan blok kode hanya jika kondisinya terpenuhi. Pada program IPK, pesan kelulusan hanya tampil jika IPK lebih dari 2,75, sedangkan baris "Program Selesai" tetap tampil karena berada di luar blok if.

3. Pernyataan if-else menyediakan jalur alternatif, sehingga selalu ada satu dari dua blok yang dijalankan. Pada program IPK versi kedua, jika kondisi tidak terpenuhi maka program menampilkan "Anda Tidak Lulus".

4. Pernyataan if-elif-else digunakan untuk kondisi yang jumlahnya lebih dari dua dan saling lepas, seperti konversi nilai angka menjadi huruf. Kondisi diperiksa dari atas ke bawah dan berhenti begitu ada yang terpenuhi, sehingga hanya satu blok yang dijalankan. Urutan kondisi harus dari batas terbesar ke terkecil agar hasilnya benar.

5. Multi if memeriksa setiap if secara independen, sehingga lebih dari satu blok dapat dijalankan sekaligus. Pada program diskon belanja, diskon dari total belanja, status member, dan kode promo dapat terakumulasi hingga 30%. Pada program SIM, multi if dipakai dengan kombinasi kondisi umur dan kepemilikan SIM sehingga hanya satu pesan yang tampil.

6. Perbedaan utama if-elif-else dan multi if terletak pada cara pemeriksaan kondisinya. if-elif-else berhenti setelah satu kondisi benar, sedangkan multi if terus memeriksa sampai if terakhir. Karena itu, pemilihan struktur harus disesuaikan dengan permasalahan agar hasil program tidak keliru.

7. match case menjadi alternatif yang lebih rapi dan mudah dibaca dibandingkan elif yang panjang untuk mencocokkan satu nilai dengan banyak pilihan tertentu. Pada program konversi bulan, case _ menangani angka di luar 1 sampai 12 dengan pesan "Angka tidak valid".

8. Fungsi input() menerima data dalam bentuk string, sehingga harus dikonversi dengan int() atau float() sebelum dibandingkan secara numerik.
