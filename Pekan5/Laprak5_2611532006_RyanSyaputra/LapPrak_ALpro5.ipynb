## 2.1 Dasar Teori


### 2.1 Dasar Teori

Perulangan merupakan salah satu struktur dasar dalam pemrograman yang digunakan untuk menjalankan suatu perintah atau sekumpulan perintah secara berulang. Perulangan dapat mengurangi penulisan kode yang sama secara berulang sehingga program menjadi lebih singkat, terstruktur, dan efisien.

Dalam bahasa pemrograman Python, salah satu jenis perulangan yang dapat digunakan adalah perulangan `for`. Perulangan `for` digunakan untuk mengulangi suatu blok kode berdasarkan setiap elemen yang terdapat pada suatu objek yang dapat diiterasi (iterable), seperti list, string, tuple, maupun hasil dari fungsi `range()`.

Bentuk umum perulangan `for` dalam Python adalah:

```python
for variabel in objek:
    perintah
```

Pada struktur tersebut, `variabel` digunakan untuk menyimpan nilai dari setiap elemen yang sedang diproses, sedangkan `objek` merupakan kumpulan data atau objek yang akan dilakukan perulangan. Perintah yang berada di dalam blok `for` akan dijalankan untuk setiap elemen yang diproses.

Salah satu fungsi yang sering digunakan bersama perulangan `for` adalah `range()`. Fungsi `range()` digunakan untuk menghasilkan urutan bilangan yang dapat digunakan sebagai jumlah atau batas perulangan. Bentuk penggunaannya antara lain `range(stop)`, `range(start, stop)`, dan `range(start, stop, step)`.

Contoh sederhana penggunaan `for` dengan `range()` adalah:

```python
for i in range(1, 6):
    print(i)
```

Program tersebut akan menampilkan angka 1 sampai 5. Angka 6 tidak ditampilkan karena nilai `stop` pada `range()` tidak termasuk dalam urutan yang dihasilkan.

Dengan memahami perulangan `for` dan fungsi `range()`, programmer dapat membuat program yang membutuhkan proses berulang dengan lebih mudah dan efisien. Konsep ini juga menjadi dasar untuk membuat berbagai program yang lebih kompleks, seperti mencetak pola, mengolah data dalam jumlah banyak, dan melakukan perhitungan secara berulang.

### 2.2.1 Program "jumlah_genap_2611532006"


````python
ulang_2006 = int(input("Masukkan nilai batas: "))

jumlah_2006 = 0
for i_2006 in range(1, ulang_2006 + 1):
    if i_2006 % 2 == 0:
        print(i_2006, end=" ")
        jumlah_2006 = jumlah_2006 + i_2006

        if i_2006 < ulang_2006   :
            print("+ ", end="")
        else:
            print("= ", jumlah_2006, end="")
print()
print("Jumlah =", jumlah_2006)
````

**Penjelasan :**

* `ulang_2006` digunakan untuk menyimpan nilai batas yang dimasukkan oleh pengguna.
* `int(input())` digunakan untuk menerima input berupa bilangan bulat.
* `jumlah_2006 = 0` digunakan sebagai nilai awal untuk menyimpan hasil penjumlahan bilangan genap.
* `for i_2006 in range(1, ulang_2006 + 1)` digunakan untuk melakukan perulangan dari angka 1 sampai nilai batas yang dimasukkan.
* `if i_2006 % 2 == 0` digunakan untuk memeriksa apakah angka yang sedang diproses merupakan bilangan genap.
* `print(i_2006, end=" ")` digunakan untuk menampilkan bilangan genap pada satu baris.
* `jumlah_2006 = jumlah_2006 + i_2006` digunakan untuk menambahkan setiap bilangan genap ke dalam total jumlah.
* `if i_2006 < ulang_2006` digunakan untuk menentukan apakah tanda `+` perlu ditampilkan.
* `print("+ ", end="")` digunakan untuk menampilkan tanda penjumlahan tanpa berpindah baris.
* `else` digunakan ketika bilangan yang diproses merupakan bilangan genap terakhir dalam batas yang ditentukan.
* `print("= ", jumlah_2006, end="")` digunakan untuk menampilkan tanda `=` dan hasil penjumlahan.
* `print()` digunakan untuk berpindah ke baris baru.
* `print("Jumlah =", jumlah_2006)` digunakan untuk menampilkan hasil akhir dari seluruh penjumlahan bilangan genap.


### 2.2.2 Program "nested_for1_2611532006"


````python
batas_2006 = int(input("Masukkan nilai batas: "))
for line_2006 in range(1, batas_2006 + 1):
    for j_2006 in range(1, (-1 * line_2006 + batas_2006) + 1):
        print(".", end="")
    print(line_2006)
````

**Penjelasan :**

* `batas_2006` digunakan untuk menyimpan nilai batas yang dimasukkan oleh pengguna.
* `int(input())` digunakan untuk menerima input berupa bilangan bulat.
* `for line_2006 in range(1, batas_2006 + 1)` digunakan untuk melakukan perulangan dari angka 1 sampai nilai batas.
* `line_2006` menunjukkan nomor baris yang sedang diproses.
* `for j_2006 in range(1, (-1 * line_2006 + batas_2006) + 1)` merupakan perulangan di dalam perulangan pertama yang digunakan untuk menentukan jumlah titik yang akan dicetak.
* `-1 * line_2006 + batas_2006` digunakan untuk mengurangi jumlah titik pada setiap baris.
* `print(".", end="")` digunakan untuk mencetak titik tanpa berpindah ke baris baru.
* `print(line_2006)` digunakan untuk mencetak nomor baris setelah seluruh titik pada baris tersebut selesai dicetak, kemudian berpindah ke baris berikutnya.
* Perulangan pertama berfungsi mengatur jumlah baris, sedangkan perulangan kedua mengatur jumlah titik pada setiap baris.
* Semakin besar nilai `line_2006`, jumlah titik yang dicetak semakin sedikit.

* `j_2006` berfungsi sebagai penghitung jumlah tanda `*` pada setiap baris.
* `print("*", end="")` digunakan untuk mencetak tanda `*` tanpa berpindah ke baris baru.
* `print()` digunakan untuk berpindah ke baris berikutnya setelah jumlah `*` pada satu baris selesai dicetak.
* Karena kedua perulangan menggunakan batas yang sama, jumlah baris dan jumlah `*` pada setiap baris akan sama.

### 2.2.4 Program "nested_for3_2611532006"


````python
batas_2006 = int(input("Masukkan nilai batas: "))
for i_2006 in range(batas_2006+1):
    for j_2006 in range(batas_2006+1):
        print(i_2006+j_2006, end=" ")
    print()  # pindah ke baris berikutnya
````

**Penjelasan :**
* `batas_2006` digunakan untuk menyimpan nilai batas yang dimasukkan oleh pengguna.
* `int(input())` digunakan untuk menerima input berupa bilangan bulat.

* `i_2006` berfungsi sebagai variabel penghitung untuk baris.
* `for j_2006 in range(batas_2006 + 1)` digunakan untuk melakukan perulangan pada setiap kolom, dimulai dari `0` sampai nilai batas.
* `j_2006` berfungsi sebagai variabel penghitung untuk kolom.
* `print(i_2006 + j_2006, end=" ")` digunakan untuk mencetak hasil penjumlahan nilai `i_2006` dan `j_2006` tanpa berpindah ke baris baru.
* Nilai yang dicetak akan berubah sesuai dengan nilai `i_2006` dan `j_2006` pada setiap proses perulangan.
* `print()` digunakan untuk berpindah ke baris berikutnya setelah seluruh kolom pada satu baris selesai diproses.
* Perulangan pertama (`i_2006`) mengatur **baris**, sedangkan perulangan kedua (`j_2006`) mengatur **kolom**.
* Program menghasilkan pola angka berdasarkan **hasil penjumlahan indeks baris dan indeks kolom**.


### 2.2.5 Program "nested_for4_2611532006"


````python
tinggi_2006 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_2006 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2006 = tinggi_2006
    c_2006 = a_2006
    lebar_2006 = (2 * tinggi_2006) - 2

    for i_2006 in range(1, tinggi_2006 + 1):
        b_2006 = c_2006 + 1

        for j_2006 in range(1, lebar_2006 + 1):

            # Baris atas dan bawah
            if i_2006 == 1 or i_2006 == tinggi_2006:
                if j_2006 == 1 or j_2006 == lebar_2006:
                    print("#", end="")
                else:
                    print("=", end="")
                    # Baris isi
            else:
                if j_2006 == 1 or j_2006 == lebar_2006:
                    print("|", end="")
                else:
                    if j_2006 == c_2006:
                        print("<", end="")
                    elif j_2006 == b_2006:
                        print(">", end="")
                    elif j_2006 == (lebar_2006 - c_2006):
                        print("<", end="")
                    elif j_2006 == (lebar_2006 - c_2006 + 1):
                        print(">", end="")
                    elif j_2006 > b_2006 and j_2006 < (lebar_2006 - c_2006):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_2006 -= 2

        if a_2006 <= 0:
            c_2006 = (-a_2006) + 2
        else:
            c_2006 = a_2006
````

**Penjelasan:**

* `tinggi_2006` digunakan untuk menyimpan tinggi pola yang dimasukkan oleh pengguna.
* `int(input())` digunakan untuk menerima input berupa bilangan bulat.
* `if tinggi_2006 % 2 != 0` digunakan untuk memeriksa apakah nilai tinggi merupakan bilangan ganjil.
* `print("Tinggi harus bilangan genap!")` menampilkan pesan jika nilai yang dimasukkan bukan bilangan genap.
* `else` digunakan untuk menjalankan pembuatan pola apabila nilai tinggi merupakan bilangan genap.
* `a_2006 = tinggi_2006` digunakan sebagai nilai awal untuk mengatur perubahan posisi pola.
* `c_2006 = a_2006` digunakan untuk menyimpan nilai posisi yang digunakan dalam menentukan letak karakter pola.
* `lebar_2006 = (2 * tinggi_2006) - 2` digunakan untuk menentukan lebar pola berdasarkan nilai tinggi yang dimasukkan.
* `for i_2006 in range(1, tinggi_2006 + 1)` digunakan untuk melakukan perulangan sebanyak tinggi pola dan mengatur setiap baris yang akan dibuat.
* `b_2006 = c_2006 + 1` digunakan untuk menentukan posisi karakter `>` berdasarkan nilai `c_2006`.
* `for j_2006 in range(1, lebar_2006 + 1)` digunakan untuk melakukan perulangan pada setiap posisi atau kolom dalam satu baris.
* `if i_2006 == 1 or i_2006 == tinggi_2006` digunakan untuk menentukan apakah baris yang sedang dibuat merupakan baris paling atas atau paling bawah.
* Pada baris paling atas dan bawah, posisi pertama dan terakhir dicetak menggunakan karakter `#`.
* Posisi lainnya pada baris paling atas dan bawah dicetak menggunakan karakter `=`.
* Pada baris bagian dalam, posisi pertama dan terakhir dicetak menggunakan karakter `|`.
* `if j_2006 == c_2006` digunakan untuk menentukan posisi karakter `<` berdasarkan nilai `c_2006`.
* `elif j_2006 == b_2006` digunakan untuk menentukan posisi karakter `>` setelah karakter `<`.
* `elif j_2006 == (lebar_2006 - c_2006)` dan `elif j_2006 == (lebar_2006 - c_2006 + 1)` digunakan untuk menentukan pasangan karakter `<` dan `>` pada sisi lainnya.
* `elif j_2006 > b_2006 and j_2006 < (lebar_2006 - c_2006)` digunakan untuk mencetak karakter `.` pada bagian tengah pola.
* `else` digunakan untuk mencetak spasi pada posisi yang tidak memenuhi kondisi sebelumnya.
* `print()` digunakan untuk berpindah ke baris berikutnya setelah seluruh posisi pada satu baris selesai diproses.
* `a_2006 -= 2` digunakan untuk mengurangi nilai `a_2006` sebanyak 2 pada setiap pergantian baris.
* `if a_2006 <= 0` digunakan untuk memeriksa apakah nilai `a_2006` sudah mencapai atau melewati 0.
* `c_2006 = (-a_2006) + 2` digunakan untuk mengatur kembali nilai `c_2006` ketika `a_2006` sudah mencapai 0 atau kurang.
* `else: c_2006 = a_2006` digunakan untuk memberikan nilai `a_2006` kepada `c_2006` selama nilainya masih lebih dari 0.
* Secara keseluruhan, program digunakan untuk **membuat pola karakter dengan ukuran tertentu berdasarkan tinggi yang dimasukkan pengguna**, dengan syarat tinggi pola harus berupa bilangan genap.

### 2.2.8 Program "perulangan_for3_2611532006"


````python
ulang_2006 = int(input("Masukkan jumlah perulangan: "))

jumlah_2006 = 0
for i_2006 in range(1, ulang_2006 + 1):
    print(i_2006, end=" ")
    jumlah_2006 = jumlah_2006 + i_2006

    if i_2006 < ulang_2006:
        print("+ ", end="")
    else:
        print("= ", jumlah_2006, end="")
print()
print("Jumlah =", jumlah_2006)
````

**Penjelasan:**
* `ulang_2006` digunakan untuk menyimpan jumlah atau batas perulangan yang dimasukkan oleh pengguna.
* `int(input())` digunakan untuk menerima input berupa bilangan bulat.
* `jumlah_2006 = 0` digunakan sebagai nilai awal untuk menyimpan hasil penjumlahan.
* `for i_2006 in range(1, ulang_2006 + 1)` digunakan untuk melakukan perulangan dari angka `1` sampai nilai `ulang_2006`.
* `print(i_2006, end=" ")` digunakan untuk menampilkan angka yang sedang diproses tanpa berpindah ke baris baru.
* `jumlah_2006 = jumlah_2006 + i_2006` digunakan untuk menambahkan setiap angka ke dalam variabel `jumlah_2006`.
* `if i_2006 < ulang_2006` digunakan untuk memeriksa apakah angka yang sedang diproses belum mencapai angka terakhir.
* `print("+ ", end="")` digunakan untuk menampilkan tanda `+` jika masih terdapat angka berikutnya.
* `else` digunakan ketika `i_2006` sudah mencapai nilai terakhir dari perulangan.
* `print("= ", jumlah_2006, end="")` digunakan untuk menampilkan tanda `=` beserta hasil penjumlahan.
* `print()` digunakan untuk berpindah ke baris berikutnya setelah seluruh proses selesai.
* `print("Jumlah =", jumlah_2006)` digunakan untuk menampilkan hasil akhir penjumlahan.

### 2.2.9 Program "latihan5_2611532006"


````python
tinggi_2006 = int(input("Masukkan tingi segitiga:"))

for i_2006 in range(1, tinggi_2006 + 1):
    print(" " * (tinggi_2006 - i_2006), end=" ")

    for j_2006 in range(i_2006):
        print("*", end=" ")
    print()  
````

**Penjelasan:**
* `tinggi_2006` digunakan untuk menyimpan tinggi segitiga yang dimasukkan oleh pengguna.
* `int(input())` digunakan untuk menerima input berupa bilangan bulat.
* `for i_2006 in range(1, tinggi_2006 + 1)` digunakan untuk melakukan perulangan dari baris pertama sampai baris terakhir sesuai tinggi segitiga.
* `i_2006` berfungsi sebagai penghitung atau nomor baris yang sedang dibuat.
* `print(" " * (tinggi_2006 - i_2006), end=" ")` digunakan untuk mencetak spasi di awal setiap baris agar posisi tanda `*` membentuk segitiga yang berada di tengah.
* Jumlah spasi dihitung dari `tinggi_2006 - i_2006`, sehingga semakin ke bawah jumlah spasi semakin sedikit.
* `for j_2006 in range(i_2006)` digunakan untuk menentukan jumlah tanda `*` yang dicetak pada setiap baris.
* Jumlah perulangan `j_2006` mengikuti nilai `i_2006`, sehingga jumlah tanda `*` bertambah satu pada setiap baris.
* `print("*", end=" ")` digunakan untuk mencetak tanda `*` tanpa berpindah baris.
* `print()` digunakan untuk berpindah ke baris berikutnya setelah seluruh tanda `*` pada satu baris selesai dicetak.
* Program menghasilkan pola **segitiga sama kaki** menggunakan perulangan `for` bertingkat.
